```python
import numpy as np
from google.colab import files
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression
```

```python
def linear_regression(x,w):
  return np.dot(x,w)
```

```python
def loss_function(x,y,w):
  return np.square(linear_regression(x,w)- y).sum() / (2 * (x.size))
```

```python
def gradient_step(x, y, w, learning_rate):
    m = len(y)
    grad = (x.T @ (linear_regression(x, w) - y)) / m
    w -= learning_rate * grad
    return w
```

```python
def gradient(x, y, learning_rate, num_iter, eps):
    ones = np.ones((x.shape[0], 1))
    x = np.hstack((ones, x))

    w = np.zeros(x.shape[1])

    loss = loss_function(x, y, w)
    loss_history = [loss]

    for _ in range(num_iter):
        w = gradient_step(x, y, w, learning_rate)

        loss = loss_function(x, y, w)
        if abs(loss - loss_history[-1]) < eps:
            loss_history.append(loss)
            break

        loss_history.append(loss)

    return w, loss_history
```
