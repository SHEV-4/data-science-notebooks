```python
import matplotlib.pyplot as plt
from keras.datasets import imdb
from keras.utils import pad_sequences
from keras.models import Sequential
from keras.layers import Embedding, GRU, Dense,SimpleRNN,LSTM,Bidirectional
```

```python
MAX_FEATURES = 10000
MAX_LEN = 500
```

```python
(input_train, y_train), (input_test, y_test) = imdb.load_data(num_words=MAX_FEATURES)
```

```python
input_train = pad_sequences(input_train, maxlen=MAX_LEN)
input_test = pad_sequences(input_test, maxlen=MAX_LEN)
```

```python
model_rnn = Sequential([
    Embedding(MAX_FEATURES,32),
    SimpleRNN(32),
    Dense(1,activation="sigmoid")
])
```

```python
model_gru = Sequential([
    Embedding(MAX_FEATURES,32),
    GRU(32),
    Dense(1,activation="sigmoid")
])
```

```python
model_lstm = Sequential([
    Embedding(MAX_FEATURES,32),
    LSTM(32),
    Dense(1,activation="sigmoid")
])
```

```python
model_brnn = Sequential([
    Embedding(MAX_FEATURES,32),
    Bidirectional(LSTM(32)),
    Dense(1,activation="sigmoid")
])
```

```python
model_drnn = Sequential([
    Embedding(MAX_FEATURES,32),
    Bidirectional(LSTM(32,return_sequences=True)),
    Bidirectional(LSTM(32)),
    Dense(1,activation="sigmoid")
])
```

```python
model_rnn.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['acc'])


model_gru.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['acc'])


model_lstm.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['acc'])


model_brnn.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['acc'])


model_drnn.compile(
    optimizer='rmsprop',
    loss='binary_crossentropy',
    metrics=['acc'])
```

```python
history_rnn = model_rnn.fit(input_train, y_train, epochs=5, batch_size=128, validation_split=0.3)
```

```python
history_gru = model_gru.fit(input_train, y_train, epochs=5, batch_size=128, validation_split=0.3)
```

```python
history_lstm = model_lstm.fit(input_train, y_train, epochs=5, batch_size=128, validation_split=0.3)
```

```python
history_brnn = model_brnn.fit(input_train, y_train, epochs=5, batch_size=128, validation_split=0.3)
```

```python
history_drnn = model_drnn.fit(input_train, y_train, epochs=5, batch_size=128, validation_split=0.3)
```

```python
acc_rnn = history_rnn.history['acc']
val_acc_rnn = history_rnn.history['val_acc']
loss_rnn = history_rnn.history['loss']
val_loss_rnn = history_rnn.history['val_loss']

acc_gru = history_gru.history['acc']
val_acc_gru = history_gru.history['val_acc']
loss_gru = history_gru.history['loss']
val_loss_gru = history_gru.history['val_loss']

acc_lstm = history_lstm.history['acc']
val_acc_lstm = history_lstm.history['val_acc']
loss_lstm = history_lstm.history['loss']
val_loss_lstm = history_lstm.history['val_loss']

acc_brnn = history_brnn.history['acc']
val_acc_brnn = history_brnn.history['val_acc']
loss_brnn = history_brnn.history['loss']
val_loss_brnn = history_brnn.history['val_loss']

acc_drnn = history_drnn.history['acc']
val_acc_drnn = history_drnn.history['val_acc']
loss_drnn = history_drnn.history['loss']
val_loss_drnn = history_drnn.history['val_loss']

epochs = range(1, len(acc_rnn) + 1)
fig, axes = plt.subplots(2, 5, figsize=(25, 5))
#---RNN---
axes[0][0].set_title('Точність тренування і валідації RNN')
axes[0][0].plot(epochs, acc_rnn, "--", label='Тренувальнні', color = "blue")
axes[0][0].plot(epochs, val_acc_rnn,  label='Валідаційні', color = "blue")
axes[0][0].legend()

axes[0][1].set_title('Точність тренування і валідації GRU')
axes[0][1].plot(epochs, acc_gru, "--", label='Тренувальнні', color = "red")
axes[0][1].plot(epochs, val_acc_gru,  label='Валідаційні', color = "red")
axes[0][1].legend()

axes[0][2].set_title('Точність тренування і валідації LSTM')
axes[0][2].plot(epochs, acc_lstm, "--", label='Тренувальнні', color = "green")
axes[0][2].plot(epochs, val_acc_lstm,  label='Валідаційні', color = "green")
axes[0][2].legend()

axes[0][3].set_title('Точність тренування і валідації BRNN')
axes[0][3].plot(epochs, acc_brnn, "--", label='Тренувальнні', color = "orange")
axes[0][3].plot(epochs, val_acc_brnn,  label='Валідаційні', color = "orange")
axes[0][3].legend()

axes[0][4].set_title('Точність тренування і валідації DRNN')
axes[0][4].plot(epochs, acc_drnn, "--", label='Тренувальнні', color = "black")
axes[0][4].plot(epochs, val_acc_drnn,  label='Валідаційні', color = "black")
axes[0][4].legend()

axes[1][0].set_title('Втрати тренування і валідації RNN')
axes[1][0].plot(epochs, loss_rnn, "--", label='Тренувальнні', color = "blue")
axes[1][0].plot(epochs, val_loss_rnn,  label='Валідаційні', color = "blue")
axes[1][0].legend()

axes[1][1].set_title('Втрати тренування і валідації GRU')
axes[1][1].plot(epochs, loss_gru, "--", label='Тренувальнні', color = "red")
axes[1][1].plot(epochs, val_loss_gru,  label='Валідаційні', color = "red")
axes[1][1].legend()


axes[1][2].set_title('Втрати тренування і валідації LSTM')
axes[1][2].plot(epochs, loss_lstm, "--", label='Тренувальнні', color = "green")
axes[1][2].plot(epochs, val_loss_lstm,  label='Валідаційні', color = "green")
axes[1][2].legend()


axes[1][3].set_title('Втрати тренування і валідації BRNN')
axes[1][3].plot(epochs, loss_brnn, "--", label='Тренувальнні', color = "orange")
axes[1][3].plot(epochs, val_loss_brnn,  label='Валідаційні', color = "orange")
axes[1][3].legend()


axes[1][4].set_title('Втрати тренування і валідації DRNN')
axes[1][4].plot(epochs, loss_drnn, "--", label='Тренувальнні', color = "black")
axes[1][4].plot(epochs, val_loss_drnn,  label='Валідаційні', color = "black")
axes[1][4].legend()

plt.show()
```

```python
test_loss_rnn, test_acc_rnn = model_rnn.evaluate(input_test, y_test)
test_loss_gru, test_acc_gru = model_gru.evaluate(input_test, y_test)
test_loss_lstm, test_acc_lstm = model_lstm.evaluate(input_test, y_test)
test_loss_brnn, test_acc_brnn = model_brnn.evaluate(input_test, y_test)
test_loss_drnn, test_acc_drnn = model_drnn.evaluate(input_test, y_test)

print("Оцінка моделі RNN")
print(f'Test Loss: {format(test_loss_rnn)} \nTest Accuracy: {format(test_acc_rnn)}')
print("-------------------")
print("Оцінка моделі GRU")
print(f'Test Loss: {format(test_loss_gru)} \nTest Accuracy: {format(test_acc_gru)}')
print("-------------------")
print("Оцінка моделі LSTM")
print(f'Test Loss: {format(test_loss_lstm)} \nTest Accuracy: {format(test_acc_lstm)}')
print("-------------------")
print("Оцінка моделі BRNN")
print(f'Test Loss: {format(test_loss_brnn)} \nTest Accuracy: {format(test_acc_brnn)}')
print("-------------------")
print("Оцінка моделі DRNN")
print(f'Test Loss: {format(test_loss_drnn)} \nTest Accuracy: {format(test_acc_drnn)}')
print("-------------------")
```

```python
predictions_rnn = model_rnn.predict(input_test)

predictions_gru = model_gru.predict(input_test)

predictions_lstm = model_lstm.predict(input_test)

predictions_brnn = model_brnn.predict(input_test)

predictions_drnn = model_drnn.predict(input_test)
```
