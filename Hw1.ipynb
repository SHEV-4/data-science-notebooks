```python
# 2.4 Створіть наступний шаблон без жорсткого кодування(Hard coding) . Використовуйте лише функції Numpy і наведений нижче масив введення Input:a = np.array([1,2,3])
# Output:array([1, 1, 1, 2, 2, 2, 3, 3, 3, 1, 2, 3, 1, 2, 3, 1, 2, 3])
a = np.array([1,2,3])
first_part = np.repeat(a,3)
last_part = np.tile(a,3)
result = np.concatenate((first_part,last_part))
print(result)
```

```python
# 2.8 Обчисліть евклідову відстань між двома масивами a і b.
# a = np.array([1,2,3,4,5]) b = np.array([4,5,6,7,8])
a = np.array([1,2,3,4,5])
b = np.array([4,5,6,7,8])
c = a - b
c_square = c ** 2
total = np.sum(c_square)
result = np.sqrt(total)

print(c)
print(c_square)
print(total)
print(result)
```

```python
# 2.10 Знайдіть повторювані записи (від 2-го випадку повторення і далі) у заданому масиві і позначте їх як True. Перше повторення має бути False.
# Input np.random.seed(100) a = np.random.randint(0, 5, 10)
# [0 0 3 0 2 4 2 2 2 2] Output [False True False True False False True True True True]

np.random.seed(100)
a = np.random.randint(0, 5, 10)
result = np.ones(len(a),dtype= bool)
_,first_indices  = np.unique(a,return_index=True)
result[first_indices] = False
print(result)
```
