```python
from surprise import SVD, Dataset,SVDpp,NMF
from surprise.model_selection import cross_validate
```

```python
data = Dataset.load_builtin('ml-100k')

model_SVD = SVD()
model_SVDpp = SVDpp() 
model_NMF = NMF() 
```

```python
cross_validate(model_SVD,data,measures=['RMSE','MAE'],cv = 5,verbose=True)
```

```python
cross_validate(model_SVDpp,data,measures=['RMSE','MAE'],cv = 5,verbose=True)
```

```python
cross_validate(model_NMF,data,measures=['RMSE','MAE'],cv = 5,verbose=True)
```

Найточніша модель SVD++ - RMSE = 0.9208, MAE = 0.7220 але вона значно довше тренується і передбачає <br>
В SVD RMSE = 0.9372, MAE = 0.7389 трохи гіршою є точність але швидкість навчання і передбачення є значно швидшим <br>
В NMF RMSE = 0.9611, MAE = 0.7553 вища похибка. За часом вона є достатньо швидка<br>
