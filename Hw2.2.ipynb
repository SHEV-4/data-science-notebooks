```python
import pandas as pd
file_path = '2017_jun_final.csv'
data = pd.read_csv(file_path)
head_data = data.head()
print(head_data)
```

```python
shape_table = data.shape
print(shape_table)
```

```python
print(data.dtypes)
```

```python
missing_count = data.isnull().sum()
print(missing_count)
print(data.columns)
```

```python
data_clean = data.loc[:,(data.columns == "Мова програмування")| ~data.isnull().any()]

print(data_clean)
```

```python
missing_count = data_clean.isnull().sum()
print(missing_count)
```

```python
data_no_nan = data_clean.dropna()
print(data_no_nan)
```

```python
print(data_no_nan.shape)
```

```python
python_data = data_no_nan[data_no_nan["Мова програмування"] == "Python"]
print(python_data)
```

```python
print(python_data.shape)
```

```python
grouped_by_post = python_data.groupby("Посада")
print(grouped_by_post.head())
```

```python
agg_salary = grouped_by_post.agg({"Зарплата на місяць": ["min", "max"]})
print(agg_salary)
```

```python
def fill_avg_salary(row):
    return python_data[python_data["Посада"] == row["Посада"]]["Зарплата на місяць"].mean()

python_data["avg"] = python_data.apply(fill_avg_salary, axis=1)
print(python_data)
```

```python
python_data["avg"].describe()
```

```python
python_data.to_csv("python_data_cleaned.csv", index=False)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
sns.boxplot(data=python_data, x="Посада", y="salary")
plt.xticks(rotation=45)
plt.title("Розподіл зарплат за посадою")
plt.show()
```

```python
avg_salary_city = python_data.groupby("Місто")["salary"].mean().sort_values(ascending=False)

avg_salary_city.plot(kind="bar", color="skyblue")
plt.ylabel("Середня зарплата")
plt.title("Середня зарплата по містах")
plt.xticks(rotation=45)
plt.show()
```

```python
plt.scatter(python_data["exp"], python_data["salary"], alpha=0.5, color="purple")
plt.xlabel("Досвід роботи (роки)")
plt.ylabel("Зарплата")
plt.title("Залежність зарплати від досвіду")
plt.grid(True)
plt.show()
```
